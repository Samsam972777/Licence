```python
# On importe Tkinter pour pouvoir créer une fenêtre
from tkinter import *

# On importe ttk pour utiliser des éléments graphiques
from tkinter import ttk

# On crée la fenêtre principale du programme
fen = Tk()

# On définit la taille de la fenêtre : 800 pixels de largeur et 600 de hauteur
fen.geometry("800x600")

# On donne un titre à la fenêtre
fen.title("Travail sur l'interface graphique")

# On crée un cadre dans la fenêtre pour ranger les éléments
frm = ttk.Frame(fen)

# On place le cadre dans la fenêtre
frm.grid()

# On crée un texte qui sera affiché dans la fenêtre
objet1 = ttk.Label(frm, text="Ceci est le premier texte")

# On place le texte dans la première colonne et la première ligne
# padx et pady permettent de laisser un espace autour du texte
objet1.grid(column=0, row=0, padx=50, pady=50)

# On crée une zone dans laquelle l'utilisateur peut écrire un nombre
objet2 = ttk.Entry(frm)

# On place la zone de saisie dans la deuxième colonne
objet2.grid(column=1, row=0)

# On crée une fonction qui sera exécutée quand on cliquera sur le bouton
def recup():

    # On récupère ce que l'utilisateur a écrit dans la zone de saisie
    # float() transforme le texte en nombre décimal
    # On ajoute 3 au nombre récupéré
    # ttk.Label() crée un nouveau texte qui affichera le résultat
    objet3 = ttk.Label(frm, text=float(objet2.get())+3)

    # On affiche le résultat dans la deuxième colonne et la troisième ligne
    objet3.grid(column=1, row=2)

# On crée un bouton avec le texte "Valider"
# Quand on clique dessus, la fonction recup() se lance
btn1 = ttk.Button(frm, text="Valider", command=recup)

# On place le bouton dans la deuxième colonne et la deuxième ligne
btn1.grid(column=1, row=1)

# On lance la fenêtre et on la laisse ouverte
fen.mainloop()

#Ce programme crée une fenêtre dans laquelle l'utilisateur peut saisir un nombre, 
#cliquer sur le bouton « Valider » et obtenir automatiquement le résultat de ce nombre additionné à 3.
```

```python
# Programme Prof simplifier par LIA. ↓ 
from tkinter import *
from tkinter import ttk

# On crée la fenêtre principale
fen = Tk()
fen.geometry("800x800")
fen.title("Calcul moyenne")

# On crée la liste des ressources
ressources = [f"R{i}" for i in range(101, 116)]

# On crée la liste des SAE
SAE = [f"SAE{i}" for i in range(11, 17)]

# Liste qui va contenir toutes les zones de saisie
entrees = []

# On crée les listes des coefficients
coef1 = [10,10,7,7,0,5,0,6,0,5,4,2,5,5,0,20,20,0,0,0,7]
coef2 = [4,0,2,8,6,0,0,0,0,5,5,2,9,9,3,0,0,29,0,0,7]
coef3 = [4,0,2,0,0,5,15,6,4,5,5,2,0,0,3,0,0,0,20,20,7]

# On crée un cadre pour organiser les éléments
frm = ttk.Frame(fen)
frm.grid()

# Fonction pour afficher les ressources et les SAE
def afficher_notes(colonne, liste):

    for i, nom in enumerate(liste):

        # On affiche le nom de la ressource ou de la SAE
        etiquette = ttk.Label(frm, text=nom)
        etiquette.grid(column=colonne, row=i, padx=10, pady=10)

        # On crée une zone pour saisir la note
        saisie = ttk.Entry(frm, width=10)
        saisie.grid(column=colonne+1, row=i, padx=10, pady=10)

        # On ajoute la zone de saisie dans la liste
        entrees.append(saisie)

# On affiche les ressources et les SAE
afficher_notes(0, ressources)
afficher_notes(2, SAE)

# Fonction pour calculer les moyennes
def recup():

    # On récupère toutes les notes
    notes = []

    for entree in entrees:
        notes.append(float(entree.get()))

    # On calcule les moyennes avec les coefficients
    moyenne1 = sum(notes[i] * coef1[i] for i in range(21)) / sum(coef1)
    moyenne2 = sum(notes[i] * coef2[i] for i in range(21)) / sum(coef2)
    moyenne3 = sum(notes[i] * coef3[i] for i in range(21)) / sum(coef3)

    # On affiche les trois résultats
    resultat.config(text="Moyenne 1 : " + str(round(moyenne1, 2))
                         + "\nMoyenne 2 : " + str(round(moyenne2, 2))
                         + "\nMoyenne 3 : " + str(round(moyenne3, 2)))

# On crée le bouton Valider
btn1 = ttk.Button(frm, text="Valider", command=recup)
btn1.grid(column=1, row=16, padx=10, pady=20)

# On crée un texte pour afficher les résultats
resultat = ttk.Label(frm, text="Moyennes : ")
resultat.grid(column=2, row=16, padx=10, pady=20)

# On lance la fenêtre
fen.mainloop()

#Le programme sert à créer une interface graphique permettant de saisir les notes de toutes les ressources et SAE d'un semestre, 
#puis de calculer automatiquement trois moyennes pondérées grâce aux coefficients propres à chaque UE.
```

```python
#Programme de la Prof ↓ 


# On importe les outils nécessaires
from tkinter import *
from tkinter import ttk
import numpy as np

# On crée la fenêtre
fen = Tk()
fen.title("Résultats du semestre 1")

# On crée un cadre pour organiser les éléments
frm = ttk.Frame(fen)
frm.grid()

# On crée la liste des matières
matieres = [f"R{i}" for i in range(101, 116)] + [f"SAE{i}" for i in range(11, 17)]

# On affiche le titre
titre = ttk.Label(frm, text="Saisie des notes")
titre.grid(column=1, row=0, padx=50, pady=20)

# On crée une liste pour stocker les zones de saisie
saisies = []

# On définit les coefficients des trois UE
coef1 = [10,10,7,7,0,5,0,6,0,5,4,2,5,5,0,20,20,0,0,0,7]
coef2 = [4,0,2,8,6,0,0,0,0,5,5,2,9,9,3,0,0,29,0,0,7]
coef3 = [4,0,2,0,0,5,15,6,4,5,5,2,0,0,3,0,0,0,20,20,7]

# On crée une zone de saisie pour chaque matière
for i, matiere in enumerate(matieres):

    mat = ttk.Label(frm, text=matiere)
    mat.grid(column=0, row=i+1, padx=20, pady=10)

    data = ttk.Entry(frm)
    data.grid(column=1, row=i+1)

    saisies.append(data)

# Fonction pour récupérer les notes et calculer les moyennes
def recup():

    # On récupère toutes les notes saisies
    notes = [float(data.get()) for data in saisies]

    # Calcul et affichage de la moyenne de l'UE1
    UE1 = ttk.Label(frm, text="Moyenne de l'UE1")
    UE1.grid(column=2, row=3, pady=20)

    moy1 = ttk.Label(frm, text=round(np.average(notes, weights=coef1), 3), relief=GROOVE)
    moy1.grid(column=3, row=3, pady=20)

    # Calcul et affichage de la moyenne de l'UE2
    UE2 = ttk.Label(frm, text="Moyenne de l'UE2")
    UE2.grid(column=2, row=5, pady=20)

    moy2 = ttk.Label(frm, text=round(np.average(notes, weights=coef2), 3), relief=GROOVE)
    moy2.grid(column=3, row=5, pady=20)

    # Calcul et affichage de la moyenne de l'UE3
    UE3 = ttk.Label(frm, text="Moyenne de l'UE3")
    UE3.grid(column=2, row=7, pady=20)

    moy3 = ttk.Label(frm, text=round(np.average(notes, weights=coef3), 3), relief=GROOVE)
    moy3.grid(column=3, row=7, pady=20)

# On crée le bouton qui lance le calcul
btn_val = ttk.Button(frm, text="Valider la saisie", command=recup)
btn_val.grid(column=2, row=1, padx=30)

# On lance la fenêtre
fen.mainloop()

#Le programme sert à créer une interface graphique permettant de saisir les notes de toutes les ressources et SAE d'un semestre, 
#puis de calculer automatiquement trois moyennes pondérées grâce aux coefficients propres à chaque UE.
```
