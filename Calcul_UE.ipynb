Version 1 (notes et coefficients au clavier)

```python
# On met toutes les variable à 0.
moy_1=0
coe_1=0
som_1=0
moy_2=0  
coe_2=0
som_2=0
moy_3=0  
coe_3=0
som_3=0

# On demande la note et les coefficients pour cheque UE, puis on met à jour les totaux
for i in range(101,116):
   note=float(input("Donnez-moi la note de la ressource R.{}".format(i)))
   coeff=int(input("Donnez-moi le coefficient de la ressource R.{} dans l'UE1".format(i)))
   som_1+=note*coeff
   coe_1+=coeff
   coeff=int(input("Donnez-moi le coefficient de la ressource R.{} dans l'UE2".format(i)))   
   som_2+=note*coeff
   coe_2+=coeff
   coeff=int(input("Donnez-moi le coefficient de la ressource R.{} dans l'UE3".format(i)))   
   som_3+=note*coeff
   coe_3+=coeff

# Pareil que pour les ressources, mais pour les SAE.
for i in range(11,17):
   note=float(input("Donnez-moi la note de la SAE.{}".format(i)))
   coeff=int(input("Donnez-moi le coefficient de la SAE.{} dans l'UE1".format(i)))
   som_1+=note*coeff
   coe_1+=coeff
   coeff=int(input("Donnez-moi le coefficient de la SAE.{} dans l'UE2".format(i)))  
   som_2+=note*coeff
   coe_2+=coeff
   coeff=int(input("Donnez-moi le coefficient de la SAE.{} dans l'UE3".format(i)))   
   som_3+=note*coeff
   coe_3+=coeff
moy_1=som_1/coe_1
moy_2=som_2/coe_2  
moy_3=som_3/coe_3 

# calcul les moyennes des 3 UE et on affiche les resultats arrondis
print("UE1 :", round(moy_1,3))
print("UE2 :", round(moy_2,3))
print("UE3 :", round(moy_3,3))
```

```python
# Importer pickel → enregistre de données askip
import pickle
# lites avec noms des matières et coefs
matiere=[f"R{i}" for i in range(101,116)] + [f"SAE{i}" for i in range(11,17)]
coef1=[10,10,7,7,0,5,0,6,0,5,4,2,5,5,0,20,20,0,0,0,7]
coef2=[4,0,2,8,6,0,0,0,0,5,5,2,9,9,3,0,0,29,0,0,7]
coef3=[4,0,2,0,0,5,15,6,4,5,5,2,0,0,3,0,0,0,20,20,7]

# Fonction pour change une list et dictio..
def dico(list_valeurs):
   return {cle : valeur for cle, valeur in zip (matiere,list_valeurs)}
# dictionnaire pour chaque UE
dico_coef1 = dico(coef1)
dico_coef2 = dico(coef2)
dico_coef3 = dico(coef3)

# on enregistre le dictionnaire dans le fichier coef_UE1
with open('coef_UE1','wb') as fichier:
   mon_objet_pickler=pickle.Pickler(fichier)
   mon_objet_pickler.dump(dico_coef1)
# on enregistre le dictionnaire dans le fichier coef_UE2
with open('coef_UE2','wb') as fichier:
   mon_objet_pickler=pickle.Pickler(fichier)
   mon_objet_pickler.dump(dico_coef2)
# on enregistre le dictionnaire dans le fichier coef_UE3
with open('coef_UE3','wb') as fichier:
   mon_objet_pickler=pickle.Pickler(fichier)
   mon_objet_pickler.dump(dico_coef3)
    
print("fichier enregistré")
```

```python
import pickle
# Lecture fichiers pour récup les dictionnaires des coefficients enregistré
with open('coef_UE1','rb') as fichier:
    mon_objet_unpickler=pickle.Unpickler(fichier)
    coef_lu1=mon_objet_unpickler.load()

with open('coef_UE2','rb') as fichier:
    mon_objet_unpickler=pickle.Unpickler(fichier)
    coef_lu2=mon_objet_unpickler.load()

with open('coef_UE3','rb') as fichier:
    mon_objet_unpickler=pickle.Unpickler(fichier)
    coef_lu3=mon_objet_unpickler.load()
# On affiche les coefficients de chaque UE..
print(coef_lu1)
print(coef_lu2)
print(coef_lu3)
#création d'une list de notes et d'un ditionnaire
list_notes = [i for i in range(21)]
dico_notes = dico(list_notes)
#initialisation
somme = 0 
total_coef1 = 0
# boucle → afficher et claculer moy UE1
for element in matiere:
    somme += coef_lu1[element]*dico_notes[element]
    total_coef1 += coef_lu1[element]
moy1 = round(somme/total_coef1, 3)
# Affiche moyenne UE1
print(f"La moyenne de l'UE1 est {moy1}")

somme = 0 
total_coef2 = 0
#calcul moyenne UE2
for element in matiere:
    somme += coef_lu2[element]*dico_notes[element]
    total_coef2 += coef_lu2[element]
moy2 = round(somme/total_coef2, 3)
# get moy UE2
print(f"La moyenne de l'UE2 est {moy2}")

somme = 0 
total_coef3 = 0
#Calcule moyennes UE3
for element in matiere:
    somme += coef_lu3[element]*dico_notes[element]
    total_coef3 += coef_lu3[element]
moy3 = round(somme/total_coef3, 3)
# afficher moy UE3
print(f"La moyenne de l'UE3 est {moy3}")
```
