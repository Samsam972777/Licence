Version 1 (notes et coefficients au clavier)

```python
# on met les sommes et les coefficients à 0 pour chaque UE
som_1=0
coe_1=0
som_2=0
coe_2=0
som_3=0
coe_3=0

# ressources R101 à R115
for i in range(101,116):
    note=float(input("Donnez-moi la note de la ressource R.{}".format(i)))
    coeff=int(input("Donnez-moi le coefficient de la ressource R.{} dans l'UE1".format(i)))
    som_1+=note*coeff   # on ajoute note x coef
    coe_1+=coeff        # on ajoute le coef
    coeff=int(input("Donnez-moi le coefficient de la ressource R.{} dans l'UE2".format(i)))
    som_2+=note*coeff
    coe_2+=coeff
    coeff=int(input("Donnez-moi le coefficient de la ressource R.{} dans l'UE3".format(i)))
    som_3+=note*coeff
    coe_3+=coeff

# SAE11 à SAE16
for i in range(11,17):
    note=float(input("Donnez-moi la note de la SAE.{}".format(i)))
    coeff=int(input("Donnez-moi le coefficient de la SAE.{} dans l'UE1".format(i)))
    som_1+=note*coeff
    coe_1+=coeff
    coeff=int(input("Donnez-moi le coefficient de la SAE.{} dans l'UE2".format(i)))
    som_2+=note*coeff
    coe_2+=coeff
    coeff=int(input("Donnez-moi le coefficient de la SAE.{} dans l'UE3".format(i)))
    som_3+=note*coeff
    coe_3+=coeff

# moyenne = somme des (note x coef) / somme des coef
moy_1=som_1/coe_1
moy_2=som_2/coe_2
moy_3=som_3/coe_3
print("UE1 :", round(moy_1,3))
print("UE2 :", round(moy_2,3))
print("UE3 :", round(moy_3,3))
```

Version 3 (Dictionnaire)

```python
# essai : un dictionnaire matière -> note
dico = {"R101" : 15, "R102" : 18}
dico["R101"]     # la note de R101
dico.items()     # les couples (matière, note)
dico.keys()      # les matières
dico.values()    # les notes
```

```python
matiere=[f"R{i}" for i in range(101,116)] + [f"SAE{i}" for i in range(11,17)]
list_notes = [i for i in range(21)]   # notes pour tester
coef1=[10,10,7,7,0,5,0,6,0,5,4,2,5,5,0,20,20,0,0,0,7]
coef2=[4,0,2,8,6,0,0,0,0,5,5,2,9,9,3,0,0,29,0,0,7]
coef3=[4,0,2,0,0,5,15,6,4,5,5,2,0,0,3,0,0,0,20,20,7]

# transforme une liste en dictionnaire {matière : valeur}
def dico(list_valeurs):
    return {cle : valeur for cle, valeur in zip (matiere,list_valeurs)}

dico_notes = dico(list_notes)
dico_coef1 = dico(coef1)
dico_coef2 = dico(coef2)
dico_coef3 = dico(coef3)
print(dico_coef1)
```

```python
# UE1
somme = 0
total_coef1 = 0
for element in matiere:
    somme += dico_coef1[element] * dico_notes[element]   # note x coef
    total_coef1 += dico_coef1[element]                    # somme des coef
moy1 = round(somme / total_coef1, 3)
print(f"La moyenne de l'UE1 est {moy1}")

# UE2
somme = 0
total_coef2 = 0
for element in matiere:
    somme += dico_coef2[element] * dico_notes[element]
    total_coef2 += dico_coef2[element]
moy2 = round(somme / total_coef2, 3)
print(f"La moyenne de l'UE2 est {moy2}")

# UE3
somme = 0
total_coef3 = 0
for element in matiere:
    somme += dico_coef3[element] * dico_notes[element]
    total_coef3 += dico_coef3[element]
moy3 = round(somme / total_coef3, 3)
print(f"La moyenne de l'UE3 est {moy3}")
```

```python
# S2 : transforme une liste en dictionnaire avec les matières du S2
def dico_S2(list_valeurs):
    return {cle : valeur for cle, valeur in zip (matiere_S2,list_valeurs)}

dico_notes_S2 = dico_S2(list_notes_S2)
dico_coef1_S2 = dico_S2(coef1_S2)
dico_coef2_S2 = dico_S2(coef2_S2)
dico_coef3_S2 = dico_S2(coef3_S2)

# calcule la moyenne d'une UE avec les dictionnaires
def moyenne_dico(notes, coefs):
    somme = 0
    total_coef = 0
    for element in notes:
        somme += coefs[element] * notes[element]
        total_coef += coefs[element]
    return round(somme / total_coef, 3)

moy1_S2 = moyenne_dico(dico_notes_S2, dico_coef1_S2)
moy2_S2 = moyenne_dico(dico_notes_S2, dico_coef2_S2)
moy3_S2 = moyenne_dico(dico_notes_S2, dico_coef3_S2)
print(f"S2 : UE1 = {moy1_S2}, UE2 = {moy2_S2}, UE3 = {moy3_S2}")
```

Version 4 (décision du jury et graphique)

```python
# moyennes du S1 et du S2 dans des dictionnaires
moy_S1 = {"UE1": moy1, "UE2": moy2, "UE3": moy3}
moy_S2 = {"UE1": moy1_S2, "UE2": moy2_S2, "UE3": moy3_S2}

# moyenne de l'année = (S1 + S2) / 2
moy_annee = {}
for ue in moy_S1:
    moy_annee[ue] = round((moy_S1[ue] + moy_S2[ue]) / 2, 3)
print("Moyennes de l'année :", moy_annee)

# donne la décision du jury
def decision(moyennes):
    ue_validees = 0      # nombre d'UE >= 10
    ue_ratees = []       # UE < 10
    for ue in moyennes:
        if moyennes[ue] >= 10:
            ue_validees += 1
        else:
            ue_ratees.append(ue)

    if ue_validees == 3:
        return "Année validée : passage en BUT 2 autorisé."
    elif ue_validees == 2:
        ue = ue_ratees[0]     # la seule UE ratée
        if moyennes[ue] >= 8:
            return f"Année non validée, mais passage en BUT 2 autorisé. Il faudra valider l'{ue} en BUT 2."
        else:
            return f"Année non validée : passage en BUT 2 refusé ({ue} < 8). Un redoublement peut être proposé."
    else:
        return "Année non validée : passage en BUT 2 exclu (au moins 2 UE < 10)."

print("Décision du jury :", decision(moy_annee))
```

```python
# test des 4 cas de la consigne
print(decision({"UE1": 12, "UE2": 11, "UE3": 10}))    # 3 UE >= 10
print(decision({"UE1": 12, "UE2": 8.5, "UE3": 10}))   # 2 UE >= 10, la 3e entre 8 et 10
print(decision({"UE1": 12, "UE2": 7, "UE3": 10}))     # 2 UE >= 10, la 3e < 8
print(decision({"UE1": 9, "UE2": 7, "UE3": 12}))      # 2 UE < 10
```

```python
import matplotlib.pyplot as plt

ues = ["UE1", "UE2", "UE3"]

plt.figure()
plt.plot(ues, list(moy_S1.values()), "bo--")      # bleu : S1
plt.plot(ues, list(moy_S2.values()), "gs--")      # vert : S2
plt.plot(ues, list(moy_annee.values()), "k*-")    # noir : année
plt.plot(ues, [10, 10, 10], "r")                  # rouge : seuil de 10
plt.plot(ues, [8, 8, 8], "r:")                    # pointillés rouges : seuil de 8
plt.title("Moyennes des UE - BUT 1 R&T")
plt.xlabel("UE")
plt.ylabel("Moyenne / 20")
plt.show()

print("Décision du jury :", decision(moy_annee))
```

Version 5 (interface graphique avec Tkinter)

```python
from tkinter import *
from tkinter import ttk
import numpy as np

# fenêtre et cadre
fen = Tk()
fen.title("Année validée ou non ?")
frm = ttk.Frame(fen)
frm.grid()

# matières du S1 et du S2
matieres_S1 = [f"R{i}" for i in range(101, 116)] + [f"SAE{i}" for i in range(11, 17)]
matieres_S2 = [f"R{i}" for i in range(201, 215)] + [f"SAE{i}" for i in range(21, 26)]

# coefficients du S1 (UE1, UE2, UE3)
coefs_S1 = [[10,10,7,7,0,5,0,6,0,5,4,2,5,5,0,20,20,0,0,0,7],
            [4,0,2,8,6,0,0,0,0,5,5,2,9,9,3,0,0,29,0,0,7],
            [4,0,2,0,0,5,15,6,4,5,5,2,0,0,3,0,0,0,20,20,7]]

# coefficients du S2 (UE1, UE2, UE3)
coefs_S2 = [[13,9,11,7,5,3,1,0,3,10,5,2,2,2,24,0,0,19,7],
            [5,0,0,3,9,7,0,0,0,4,4,2,4,8,0,20,0,13,5],
            [5,4,0,4,0,0,7,7,8,4,4,4,6,0,0,0,20,14,6]]

# titres des colonnes
titre1 = ttk.Label(frm, text="Notes du S1")
titre1.grid(column=1, row=0, padx=10, pady=10)
titre2 = ttk.Label(frm, text="Notes du S2")
titre2.grid(column=3, row=0, padx=10, pady=10)

# cases de saisie du S1
saisies_S1 = []
for i, matiere in enumerate(matieres_S1):
    mat = ttk.Label(frm, text=matiere)
    mat.grid(column=0, row=i+1, padx=10, pady=2)
    data = ttk.Entry(frm, width=8)
    data.grid(column=1, row=i+1)
    saisies_S1.append(data)

# cases de saisie du S2
saisies_S2 = []
for i, matiere in enumerate(matieres_S2):
    mat = ttk.Label(frm, text=matiere)
    mat.grid(column=2, row=i+1, padx=10, pady=2)
    data = ttk.Entry(frm, width=8)
    data.grid(column=3, row=i+1)
    saisies_S2.append(data)

# décision du jury (la même que dans la version 4)
def decision(moyennes):
    ue_validees = 0
    ue_ratees = []
    for ue in moyennes:
        if moyennes[ue] >= 10:
            ue_validees += 1
        else:
            ue_ratees.append(ue)

    if ue_validees == 3:
        return "Année validée : passage en BUT 2 autorisé."
    elif ue_validees == 2:
        ue = ue_ratees[0]
        if moyennes[ue] >= 8:
            return f"Année non validée, mais passage en BUT 2 autorisé. Il faudra valider l'{ue} en BUT 2."
        else:
            return f"Année non validée : passage en BUT 2 refusé ({ue} < 8). Un redoublement peut être proposé."
    else:
        return "Année non validée : passage en BUT 2 exclu (au moins 2 UE < 10)."

# récupère les notes, calcule et affiche les résultats
def recup():
    notes_S1 = [float(data.get()) for data in saisies_S1]   # notes avec un point (ex : 12.5)
    notes_S2 = [float(data.get()) for data in saisies_S2]

    # titres du tableau des résultats
    ttk.Label(frm, text="S1").grid(column=5, row=2, padx=10)
    ttk.Label(frm, text="S2").grid(column=6, row=2, padx=10)
    ttk.Label(frm, text="Année").grid(column=7, row=2, padx=10)

    moy_annee = {}
    for i in range(3):
        ue = f"UE{i+1}"
        moy1 = round(np.average(notes_S1, weights=coefs_S1[i]), 2)   # moyenne S1
        moy2 = round(np.average(notes_S2, weights=coefs_S2[i]), 2)   # moyenne S2
        moy_annee[ue] = round((moy1 + moy2) / 2, 2)                  # moyenne de l'année

        ttk.Label(frm, text=f"Moyenne de l'{ue}").grid(column=4, row=i+3, padx=10, pady=10)
        ttk.Label(frm, text=moy1, relief=GROOVE).grid(column=5, row=i+3, padx=10)
        ttk.Label(frm, text=moy2, relief=GROOVE).grid(column=6, row=i+3, padx=10)
        ttk.Label(frm, text=moy_annee[ue], relief=GROOVE).grid(column=7, row=i+3, padx=10)

    # décision du jury (replace : retour à la ligne après chaque phrase)
    dec = ttk.Label(frm, text=decision(moy_annee).replace(". ", ".\n"))
    dec.grid(column=4, row=7, padx=10, pady=20)

# bouton qui lance le calcul
btn_val = ttk.Button(frm, text="Valider la saisie", command=recup)
btn_val.grid(column=4, row=1, padx=30)

fen.mainloop()
```
