```python
# ajouté : semestre 2 avec les dictionnaires (même méthode)
# dico() utilise la liste matiere, donc on fait une 2e fonction pour le S2
def dico_S2(list_valeurs):
    return {cle : valeur for cle, valeur in zip (matiere_S2,list_valeurs)}

dico_notes_S2 = dico_S2(list_notes_S2)
dico_coef1_S2 = dico_S2(coef1_S2)
dico_coef2_S2 = dico_S2(coef2_S2)
dico_coef3_S2 = dico_S2(coef3_S2)

def moyenne_dico(notes, coefs):
    somme = 0
    total_coef = 0
    for element in notes:
        somme += coefs[element] * notes[element]
        total_coef += coefs[element]
    return round(somme / total_coef, 3)

moy1_S2 = moyenne_dico(dico_notes_S2, dico_coef1_S2)
moy2_S2 = moyenne_dico(dico_notes_S2, dico_coef2_S2)
moy3_S2 = moyenne_dico(dico_notes_S2, dico_coef3_S2)
print(f"S2 : UE1 = {moy1_S2}, UE2 = {moy2_S2}, UE3 = {moy3_S2}")
```

Version 4 (décision du jury et graphique)

```python
# Version 4 : moyenne de l'année et décision du jury
# On range les moyennes du S1 (moy1, moy2, moy3) et du S2 dans des dictionnaires
moy_S1 = {"UE1": moy1, "UE2": moy2, "UE3": moy3}
moy_S2 = {"UE1": moy1_S2, "UE2": moy2_S2, "UE3": moy3_S2}

# Moyenne de l'année d'une UE = moyenne du S1 et du S2
moy_annee = {}
for ue in moy_S1:
    moy_annee[ue] = round((moy_S1[ue] + moy_S2[ue]) / 2, 3)
print("Moyennes de l'année :", moy_annee)

def decision(moyennes):
    ue_validees = 0
    ue_ratees = []
    for ue in moyennes:
        if moyennes[ue] >= 10:
            ue_validees += 1
        else:
            ue_ratees.append(ue)

    if ue_validees == 3:
        return "Année validée : passage en BUT 2 autorisé."
    elif ue_validees == 2:
        ue = ue_ratees[0]
        if moyennes[ue] >= 8:
            return f"Année non validée, mais passage en BUT 2 autorisé. Il faudra valider l'{ue} en BUT 2."
        else:
            return f"Année non validée : passage en BUT 2 refusé ({ue} < 8). Un redoublement peut être proposé."
    else:
        return "Année non validée : passage en BUT 2 exclu (au moins 2 UE < 10)."

print("Décision du jury :", decision(moy_annee))
```

```python
# On vérifie la fonction decision() avec les 4 cas de la consigne
print(decision({"UE1": 12, "UE2": 11, "UE3": 10}))    # 3 UE >= 10
print(decision({"UE1": 12, "UE2": 8.5, "UE3": 10}))   # 2 UE >= 10 et la 3e entre 8 et 10
print(decision({"UE1": 12, "UE2": 7, "UE3": 10}))     # 2 UE >= 10 et la 3e < 8
print(decision({"UE1": 9, "UE2": 7, "UE3": 12}))      # 2 UE < 10
```

Version 5 (interface graphique avec Tkinter)

```python
# Version 5 : interface graphique avec Tkinter
# (saisie des notes, lancement des calculs et affichage des résultats)
from tkinter import *
from tkinter import ttk
import numpy as np

# ---------- Données ----------
matieres_S1 = [f"R{i}" for i in range(101, 116)] + [f"SAE{i}" for i in range(11, 17)]
coef_S1 = {"UE1": [10,10,7,7,0,5,0,6,0,5,4,2,5,5,0,20,20,0,0,0,7],
           "UE2": [4,0,2,8,6,0,0,0,0,5,5,2,9,9,3,0,0,29,0,0,7],
           "UE3": [4,0,2,0,0,5,15,6,4,5,5,2,0,0,3,0,0,0,20,20,7]}

matieres_S2 = [f"R{i}" for i in range(201, 215)] + [f"SAE{i}" for i in range(21, 26)]
coef_S2 = {"UE1": [13,9,11,7,5,3,1,0,3,10,5,2,2,2,24,0,0,19,7],
           "UE2": [5,0,0,3,9,7,0,0,0,4,4,2,4,8,0,20,0,13,5],
           "UE3": [5,4,0,4,0,0,7,7,8,4,4,4,6,0,0,0,20,14,6]}

# Notes de test (pour remplir vite les cases)
test_S1 = [12.5, 11, 9.5, 8, 10.5, 13, 14.5, 9, 15, 12, 13.5, 16, 7.5, 8.5, 12, 14, 11.5, 10, 15.5, 13, 14]
test_S2 = [11, 12.5, 10, 9.5, 6, 7, 13, 12, 14, 12, 14, 15, 9.5, 5, 12.5, 7, 14, 12, 13]

def decision(moyennes):
    ue_validees = 0
    ue_ratees = []
    for ue in moyennes:
        if moyennes[ue] >= 10:
            ue_validees += 1
        else:
            ue_ratees.append(ue)

    if ue_validees == 3:
        return "Année validée : passage en BUT 2 autorisé."
    elif ue_validees == 2:
        ue = ue_ratees[0]
        if moyennes[ue] >= 8:
            return f"Année non validée, mais passage en BUT 2 autorisé. Il faudra valider l'{ue} en BUT 2."
        else:
            return f"Année non validée : passage en BUT 2 refusé ({ue} < 8). Un redoublement peut être proposé."
    else:
        return "Année non validée : passage en BUT 2 exclu (au moins 2 UE < 10)."

# ---------- Fenêtre ----------
fen = Tk()
fen.title("Année validée ou non ? - BUT 1 R&T")

frm = ttk.Frame(fen, padding=10)
frm.grid()

titre = ttk.Label(frm, text="Saisie des notes de l'année", font=("Arial", 14, "bold"))
titre.grid(column=0, row=0, columnspan=6, pady=10)

# Crée une colonne "matière + case de saisie" et renvoie la liste des cases
def creer_saisies(nom_semestre, matieres, colonne):
    ttk.Label(frm, text=nom_semestre, font=("Arial", 11, "bold")).grid(column=colonne, row=1, columnspan=2)
    saisies = []
    for i, matiere in enumerate(matieres):
        ttk.Label(frm, text=matiere).grid(column=colonne, row=i+2, padx=5, pady=2, sticky="e")
        case = ttk.Entry(frm, width=7)
        case.grid(column=colonne+1, row=i+2, padx=5, pady=2)
        saisies.append(case)
    return saisies

saisies_S1 = creer_saisies("Semestre 1", matieres_S1, 0)
saisies_S2 = creer_saisies("Semestre 2", matieres_S2, 2)

# ---------- Zone des résultats ----------
frm_res = ttk.Frame(frm, padding=10)
frm_res.grid(column=4, row=1, rowspan=22, sticky="n", padx=20)

# Tableau : UE | S1 | S2 | Année
entetes = ["", "S1", "S2", "Année"]
for j in range(4):
    ttk.Label(frm_res, text=entetes[j], font=("Arial", 10, "bold")).grid(column=j, row=2, padx=8)

labels_res = {}   # {UE : [label S1, label S2, label Année]}
for i, ue in enumerate(["UE1", "UE2", "UE3"]):
    ttk.Label(frm_res, text=ue).grid(column=0, row=i+3, padx=8, pady=3)
    labels_res[ue] = []
    for j in range(3):
        lab = ttk.Label(frm_res, text="-", width=6, relief=GROOVE, anchor="center")
        lab.grid(column=j+1, row=i+3, padx=8, pady=3)
        labels_res[ue].append(lab)

message = ttk.Label(frm_res, text="", wraplength=300, font=("Arial", 10, "bold"))
message.grid(column=0, row=7, columnspan=4, pady=10)

# Canvas pour dessiner le graphique des moyennes de l'année
graph = Canvas(frm_res, width=300, height=230, bg="white")
graph.grid(column=0, row=8, columnspan=4, pady=10)

def dessiner(moy_annee):
    graph.delete("all")
    bas = 210                 # position du bas du graphique
    echelle = 9               # 1 point = 9 pixels (20 points = 180 pixels)
    # ligne du seuil 10
    graph.create_line(20, bas - 10*echelle, 290, bas - 10*echelle, fill="green", dash=(4, 2))
    graph.create_text(270, bas - 10*echelle - 8, text="10", fill="green")
    for i, ue in enumerate(moy_annee):
        moy = moy_annee[ue]
        if moy >= 10:
            couleur = "green"
        elif moy >= 8:
            couleur = "orange"
        else:
            couleur = "red"
        x = 40 + i * 85
        graph.create_rectangle(x, bas - moy*echelle, x + 50, bas, fill=couleur)
        graph.create_text(x + 25, bas - moy*echelle - 10, text=moy)
        graph.create_text(x + 25, bas + 10, text=ue)

# Vérifie qu'un texte est bien un nombre (ex : "12.5")
def est_un_nombre(texte):
    return texte.replace(".", "", 1).isdigit()

# Récupère les notes d'une liste de cases. Renvoie None s'il y a une erreur.
def lire_notes(saisies, matieres):
    notes = []
    for i, case in enumerate(saisies):
        texte = case.get().strip().replace(",", ".")    # on accepte 12,5
        if not est_un_nombre(texte) or float(texte) > 20:
            message.config(text=f"Note incorrecte ou manquante en {matieres[i]}.", foreground="red")
            return None
        notes.append(float(texte))
    return notes

def calculer():
    notes_S1 = lire_notes(saisies_S1, matieres_S1)
    if notes_S1 is None:
        return
    notes_S2 = lire_notes(saisies_S2, matieres_S2)
    if notes_S2 is None:
        return

    moy_annee = {}
    for ue in ["UE1", "UE2", "UE3"]:
        moy1 = round(np.average(notes_S1, weights=coef_S1[ue]), 2)
        moy2 = round(np.average(notes_S2, weights=coef_S2[ue]), 2)
        moy_annee[ue] = round((moy1 + moy2) / 2, 2)
        labels_res[ue][0].config(text=moy1)
        labels_res[ue][1].config(text=moy2)
        labels_res[ue][2].config(text=moy_annee[ue])

    texte = decision(moy_annee)
    if texte.startswith("Année validée"):
        message.config(text=texte, foreground="green")
    else:
        message.config(text=texte, foreground="red")
    dessiner(moy_annee)

def remplir_test():
    for i in range(len(saisies_S1)):
        saisies_S1[i].delete(0, END)
        saisies_S1[i].insert(0, test_S1[i])
    for i in range(len(saisies_S2)):
        saisies_S2[i].delete(0, END)
        saisies_S2[i].insert(0, test_S2[i])

def effacer():
    for case in saisies_S1 + saisies_S2:
        case.delete(0, END)
    for ue in labels_res:
        for lab in labels_res[ue]:
            lab.config(text="-")
    message.config(text="")
    graph.delete("all")

# ---------- Boutons ----------
ttk.Button(frm_res, text="Calculer", command=calculer).grid(column=0, row=1, columnspan=2, pady=5)
ttk.Button(frm_res, text="Notes de test", command=remplir_test).grid(column=2, row=1, pady=5)
ttk.Button(frm_res, text="Effacer", command=effacer).grid(column=3, row=1, pady=5)

fen.mainloop()
```
