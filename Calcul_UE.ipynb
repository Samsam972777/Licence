```python
moy_1=0
coe_1=0
som_1=0
moy_2=0  
coe_2=0
som_2=0
moy_3=0  
coe_3=0
som_3=0

# modifié : R101 à R115
for i in range(101,116):
   note=float(input("Donnez-moi la note de la ressource R.{}".format(i)))
   coeff=int(input("Donnez-moi le coefficient de la ressource R.{} dans l'UE1".format(i)))
   som_1+=note*coeff
   coe_1+=coeff
   coeff=int(input("Donnez-moi le coefficient de la ressource R.{} dans l'UE2".format(i)))   # ajouté
   som_2+=note*coeff
   coe_2+=coeff
   coeff=int(input("Donnez-moi le coefficient de la ressource R.{} dans l'UE3".format(i)))   # ajouté
   som_3+=note*coeff
   coe_3+=coeff
# modifié : SAE11 à SAE16
for i in range(11,17):
   note=float(input("Donnez-moi la note de la SAE.{}".format(i)))
   coeff=int(input("Donnez-moi le coefficient de la SAE.{} dans l'UE1".format(i)))
   som_1+=note*coeff
   coe_1+=coeff
   coeff=int(input("Donnez-moi le coefficient de la SAE.{} dans l'UE2".format(i)))   # ajouté
   som_2+=note*coeff
   coe_2+=coeff
   coeff=int(input("Donnez-moi le coefficient de la SAE.{} dans l'UE3".format(i)))   # ajouté
   som_3+=note*coeff
   coe_3+=coeff
moy_1=som_1/coe_1
moy_2=som_2/coe_2  
moy_3=som_3/coe_3 
print("UE1 :", round(moy_1,3))
print("UE2 :", round(moy_2,3))
print("UE3 :", round(moy_3,3)
```

```python
import pickle

# mes listes (comme dans ma version 3)
matiere=[f"R{i}" for i in range(101,116)] + [f"SAE{i}" for i in range(11,17)]
coef1=[10,10,7,7,0,5,0,6,0,5,4,2,5,5,0,20,20,0,0,0,7]
coef2=[4,0,2,8,6,0,0,0,0,5,5,2,9,9,3,0,0,29,0,0,7]
coef3=[4,0,2,0,0,5,15,6,4,5,5,2,0,0,3,0,0,0,20,20,7]

# ma fonction qui transforme une liste en dictionnaire
def dico(list_valeurs):
   return {cle : valeur for cle, valeur in zip (matiere,list_valeurs)}

dico_coef1 = dico(coef1)
dico_coef2 = dico(coef2)
dico_coef3 = dico(coef3)

# on enregistre le dictionnaire dans le fichier coef_UE1
with open('coef_UE1','wb') as fichier:
   mon_objet_pickler=pickle.Pickler(fichier)
   mon_objet_pickler.dump(dico_coef1)

with open('coef_UE2','wb') as fichier:
   mon_objet_pickler=pickle.Pickler(fichier)
   mon_objet_pickler.dump(dico_coef2)

with open('coef_UE3','wb') as fichier:
   mon_objet_pickler=pickle.Pickler(fichier)
   mon_objet_pickler.dump(dico_coef3)

print("fichier enregistré")
```
