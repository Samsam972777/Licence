# Graphique des moyennes des UE

Calcul des moyennes des 3 UE (avec les coefficients) puis graphique en barres :
- vert : moyenne ≥ 10
- orange : entre 8 et 10
- rouge : moyenne < 8

**Avant de lancer dans Thonny :** installer matplotlib.
Menu *Outils* → *Ouvrir la console système…* puis taper `pip install matplotlib` et Entrée.

## Version 1 : notes dans une liste

```python
import matplotlib.pyplot as plt

# Les coefficients de chaque matière dans chaque UE
# (dans l'ordre : R101 à R115, puis SAE11 à SAE16)
coef1 = [10,10,7,7,0,5,0,6,0,5,4,2,5,5,0,20,20,0,0,0,7]
coef2 = [4,0,2,8,6,0,0,0,0,5,5,2,9,9,3,0,0,29,0,0,7]
coef3 = [4,0,2,0,0,5,15,6,4,5,5,2,0,0,3,0,0,0,20,20,7]

# Mes 21 notes (dans le même ordre que les coefficients)
notes = [12, 14, 9, 11, 15, 10, 8, 13, 16, 12, 9, 14, 11, 10, 15, 13, 12, 9, 11, 7, 14]


# Calcule la moyenne d'une UE avec ses coefficients
def moyenne_ue(notes, coefs):
    somme = 0
    for i in range(len(notes)):
        somme = somme + notes[i] * coefs[i]
    return round(somme / sum(coefs), 2)


# Donne la couleur de la barre selon la moyenne
def code_couleur(moy):
    if moy >= 10:
        return "green"
    elif moy < 8:
        return "red"
    else:
        return "orange"


# Calcul des 3 moyennes
moy1 = moyenne_ue(notes, coef1)
moy2 = moyenne_ue(notes, coef2)
moy3 = moyenne_ue(notes, coef3)
moyennes = [moy1, moy2, moy3]
print(f"UE1 : {moy1}   UE2 : {moy2}   UE3 : {moy3}")

# Une couleur par moyenne
couleurs = []
for moy in moyennes:
    couleurs.append(code_couleur(moy))

# Le graphique
plt.bar(["UE1", "UE2", "UE3"], moyennes, color=couleurs)
plt.title("Mes moyennes du semestre 1")
plt.ylim(0, 20)
plt.savefig("graph.png")   # on enregistre AVANT show()
plt.show()
```

## Version 2 : saisie des notes avec Tkinter

On tape les 21 notes, on clique sur **Calculer** : les moyennes s'affichent et le graphique s'ouvre.

```python
from tkinter import *
from tkinter import ttk
import matplotlib.pyplot as plt

# Les coefficients de chaque matière dans chaque UE
# (dans l'ordre : R101 à R115, puis SAE11 à SAE16)
coef1 = [10,10,7,7,0,5,0,6,0,5,4,2,5,5,0,20,20,0,0,0,7]
coef2 = [4,0,2,8,6,0,0,0,0,5,5,2,9,9,3,0,0,29,0,0,7]
coef3 = [4,0,2,0,0,5,15,6,4,5,5,2,0,0,3,0,0,0,20,20,7]

# La liste des ressources : R101 à R115
ressources = []
for i in range(101, 116):
    ressources.append(f"R{i}")

# La liste des SAE : SAE11 à SAE16
sae = []
for i in range(11, 17):
    sae.append(f"SAE{i}")


# Calcule la moyenne d'une UE avec ses coefficients
def moyenne_ue(notes, coefs):
    somme = 0
    for i in range(len(notes)):
        somme = somme + notes[i] * coefs[i]
    return round(somme / sum(coefs), 2)


# Donne la couleur de la barre selon la moyenne
def code_couleur(moy):
    if moy >= 10:
        return "green"
    elif moy < 8:
        return "red"
    else:
        return "orange"


# Dessine le graphique des 3 moyennes
def afficher_graphique(moyennes):
    couleurs = []
    for moy in moyennes:
        couleurs.append(code_couleur(moy))
    plt.close()   # ferme l'ancien graphique s'il est encore ouvert
    plt.bar(["UE1", "UE2", "UE3"], moyennes, color=couleurs)
    plt.title("Mes moyennes du semestre 1")
    plt.ylim(0, 20)
    plt.savefig("graph.png")
    plt.show()


# La fenêtre
fen = Tk()
fen.title("Moyennes du semestre 1")

frm = ttk.Frame(fen)
frm.grid()

# Liste qui garde toutes les zones de saisie
saisies = []


# Crée un Label + une zone de saisie pour chaque matière d'une liste
def creer_zones(liste, colonne):
    for i, nom in enumerate(liste):
        label = ttk.Label(frm, text=nom)
        label.grid(column=colonne, row=i, padx=10, pady=3)
        zone = ttk.Entry(frm, width=8)
        zone.grid(column=colonne + 1, row=i, padx=10, pady=3)
        saisies.append(zone)


creer_zones(ressources, 0)   # les ressources en colonnes 0 et 1
creer_zones(sae, 2)          # les SAE en colonnes 2 et 3

# Le Label qui affichera les moyennes
resultat = ttk.Label(frm, text="")
resultat.grid(column=1, row=16, padx=10)   # sous R115


# Lancée par le bouton : lit les notes, calcule, affiche
def calculer():
    notes = []
    for zone in saisies:
        texte = zone.get().replace(",", ".")   # 12,5 devient 12.5
        if texte == "":
            resultat.config(text="Il manque une note !")
            return
        notes.append(float(texte))

    moy1 = moyenne_ue(notes, coef1)
    moy2 = moyenne_ue(notes, coef2)
    moy3 = moyenne_ue(notes, coef3)

    # \n = retour à la ligne
    resultat.config(text=f"UE1 : {moy1}\nUE2 : {moy2}\nUE3 : {moy3}")
    afficher_graphique([moy1, moy2, moy3])


btn = ttk.Button(frm, text="Calculer", command=calculer)
btn.grid(column=1, row=15, padx=10, pady=10)   # sous R115

fen.mainloop()
```
