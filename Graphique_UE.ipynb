```python
# Importe le modul matplotlib pour creer graphiques 
import matplotlib.pyplot as plt
# liste des coefs 1à3
coef1 = [10,10,7,7,0,5,0,6,0,5,4,2,5,5,0,20,20,0,0,0,7]
coef2 = [4,0,2,8,6,0,0,0,0,5,5,2,9,9,3,0,0,29,0,0,7]
coef3 = [4,0,2,0,0,5,15,6,4,5,5,2,0,0,3,0,0,0,20,20,7]
# liste des notes
notes = [12, 14, 9, 11, 15, 10, 8, 13, 16, 12, 9, 14, 11, 10, 15, 13, 12, 9, 11, 7, 14]
# focntion pour calculer moyenne d'une seule UE
def moyenne_ue(notes, coefs):
    somme = 0
    for i in range(len(notes)):
        somme = somme + notes[i] * coefs[i]
    return round(somme / sum(coefs), 2)

# Création fonction pour la couleur de la barre par rapport à la moyenne
def code_couleur(moy):
    if moy >= 10:
        return "green"
    elif moy < 8:
        return "red"
    else:
        return "orange"
# Calcul et affichage les 3 moys 
moy1 = moyenne_ue(notes, coef1)
moy2 = moyenne_ue(notes, coef2)
moy3 = moyenne_ue(notes, coef3)
moyennes = [moy1, moy2, moy3]
print(f"UE1 : {moy1}   UE2 : {moy2}   UE3 : {moy3}")
# on donne une coueur à chaque moyenne pour le grphique 
couleurs = []
for moy in moyennes:
    couleurs.append(code_couleur(moy))

# et ici on affiche et génere le graphique.
plt.bar(["UE1", "UE2", "UE3"], moyennes, color=couleurs)
plt.title("Mes moyennes du semestre 1")
plt.ylim(0, 20)
plt.savefig("graph.png")   # on enregistre TOUJOURS AVANT show() → (d'apr la prof)
plt.show()
```

```python
# importe les modules pour crere l'interface graphique (tkinter) et le graphique (matplotlib)
from tkinter import *
from tkinter import ttk
import matplotlib.pyplot as plt

# liste des coefs de chaque matière pour chque UE
coef1 = [10,10,7,7,0,5,0,6,0,5,4,2,5,5,0,20,20,0,0,0,7]
coef2 = [4,0,2,8,6,0,0,0,0,5,5,2,9,9,3,0,0,29,0,0,7]
coef3 = [4,0,2,0,0,5,15,6,4,5,5,2,0,0,3,0,0,0,20,20,7]

# La lste des ressources : R101 à R115. 
ressources = []
for i in range(101, 116):
    ressources.append(f"R{i}")

# préparation liste des SAE : SAE11 à SAE16
sae = []
for i in range(11, 17):
    sae.append(f"SAE{i}")


# Calcule la moyenne d'une UE avec ses coefficients
def moyenne_ue(notes, coefs):
    somme = 0
    for i in range(len(notes)):
        somme = somme + notes[i] * coefs[i]
    return round(somme / sum(coefs), 2)


# donne la couleur de la barre selon la moyenn
def code_couleur(moy):
    if moy >= 10:
        return "green"
    elif moy < 8:
        return "red"
    else:
        return "orange"


# Dessine le graphique des 3 moyennes et on l'affiche aussi du coup
def afficher_graphique(moyennes):
    couleurs = []
    for moy in moyennes:
        couleurs.append(code_couleur(moy))
    plt.close()   # ferme l'ancien graphique s'il est encore ouvert
    plt.bar(["UE1", "UE2", "UE3"], moyennes, color=couleurs)
    plt.title("Mes moyennes du semestre 1")
    plt.ylim(0, 20)
    plt.savefig("graph.png")
    plt.show()


# La fenêtre
fen = Tk()
fen.title("Moyennes du semestre 1")
# création d'un cadre → Frame = cadre → grid = grille 
frm = ttk.Frame(fen)
frm.grid()

# Liste qui garde toutes les zones de saisie
saisies = []


# crée fonction qui affiche les label (etiquette) et la zone de saisie
def creer_zones(liste, colonne):
    for i, nom in enumerate(liste):
        label = ttk.Label(frm, text=nom)
        label.grid(column=colonne, row=i, padx=10, pady=3)
        zone = ttk.Entry(frm, width=8)
        zone.grid(column=colonne + 1, row=i, padx=10, pady=3)
        saisies.append(zone)

# on appelle les fonction ressources et sae 
creer_zones(ressources, 0)   # les ressources en colonnes 0 et 1
creer_zones(sae, 2)          # les SAE en colonnes 2 et 3

# etiquette et le réultat 
resultat = ttk.Label(frm, text="")
resultat.grid(column=1, row=16, padx=10) # padx permet de décaler vers la gauche 

#On crée la fonction qui permet de lire les notes, calculer les moys et mettre l'affichage
def calculer():
    notes = []
    for zone in saisies:
        texte = zone.get().replace(",", ".")   
        if texte == "":
            resultat.config(text="Il manque une note !")
            return
        notes.append(float(texte))

    moy1 = moyenne_ue(notes, coef1)
    moy2 = moyenne_ue(notes, coef2)
    moy3 = moyenne_ue(notes, coef3)

    resultat.config(text=f"UE1 : {moy1}\nUE2 : {moy2}\nUE3 : {moy3}") # \n = retour à la ligne
    afficher_graphique([moy1, moy2, moy3])

# Le bouton qui fait en sorte que le calcule se lance. 
btn = ttk.Button(frm, text="Calculer", command=calculer)
btn.grid(column=1, row=15, padx=10, pady=10)   
# tkinter se finit toujours comme ça. 
fen.mainloop()
```
